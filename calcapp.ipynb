```python
# Anshuk Arun
# CSC 2720
# Assignment 2

# simple calculator expression
# in-fix expression
# using lists as stacks

# Run Calculator Expression Program, given an infix expression
def calculatorExpression(infix):
        
    # curstr will be initalized with the full expression.
    # it will contain the number or operator as we go through the expression
    curstr = infix
    
    # two index pointers to keep track of current number/operator
    # i will contain where the previous seperation is
    # j will contain where the next seperation is
    i = -1
    j = 0
    
    # go through full expression. when reach the end of expression, do not need to do any more
    while j != -1:
    
        # find the seperation or " "
        j = infix.find(" ", i+1)
    
        # str is storing the current number or operator
        # no separation found, or end of expression
        if j == -1:
            curStr = infix[i+1:]            
            # if last element in expression is a number, store and do the final calculation
            if curStr.isnumeric():
                numstack.append(int(curStr))
                
                # calculate with all remaining numbers and operators in stacks
                while len(numstack) >= 2:
                    calc()
                
                # print the final result in numstack
                print("Result: ", numstack.pop())
            # last element in expression is an operator, or something else. its an invalid expression
            else:
                print("Invalid Expression")
                break
        # seperation found / still more expression afterwards
        else:
            # the number or operator is stored in str
            curstr = infix[i+1:j]
            
            # store the current space seperation as i, such that we can find the next number or operation in the next loop
            i = j
            
            # function call: determine if the curstr is a number or operator or invalid
            # then, depending on the right conditions, will call another function to do a calculation
            if numop(curstr) == -1:
                # if finding this is an invalid expression, break immediately
                print("Invalid Expression")
                break
    # clear after result or invalid has been done
    numstack.clear()
    opstack.clear()
            

# determine if the given str is a number or an operator
# with the number or operation, put in correct stack
def numop(gstr):

    # given str is a number
    if gstr.isnumeric():
        numstack.append(int(gstr))
    # given str might be an operator
    elif gstr == "+" or gstr == "-" or gstr == "*" or gstr == "/":        
        # is there enough numbers in the numstack to do an calculation?
        # YES
        if len(numstack) >= 2:

            # Determining Precedence of Top Operator over Precedence of New Operator
            # is the precedence of the top operator in the stack >= precedence of the new operator?
            grequal = False
            # chance that the top operator is not >= new operator
            if opstack[-1] == "+" or opstack[-1] == "-":
                # new operator is * or /
                if gstr == "*" or gstr == "/":
                    # top operator is less than new operator
                    grequal = False
                else:
                    grequal = True
            # if stack is * or /, then no matter what the new operator is, the top operator is >= new operator
            else:
                grequal = True

            # if top operator is >= new operator, do calculation
            if grequal:
                # calculation with previous two numbers in number stack, and the top operator stack
                calc()

                # after calculation, the previous two numbers have been removed from numstack.
                # the previous top operator has been removed from operator stack.

                # add new operator in expression to operator stack
                opstack.append(gstr)
            # top operator is < new operation, just store new operator in operator stack
            else:
                opstack.append(gstr)
        
        # NO, not enough numbers in numberstack to do any calculations
        # store operator in operator stack for now
        else:
            opstack.append(gstr)
    # given str is an invalid expression. return -1 such that program will end immediately
    else:
        return -1


# Calculation of previous two numbers and the Top Operator of stack
def calc():
    # get the operator out of stack for calculation
    op = opstack.pop()
    
    # do calculation
    # add result to numstack
    result = 0
    if op == "+":
        right = numstack.pop()
        left = numstack.pop()
        result = left + right
        numstack.append(result)
    elif op == "-":
        right = numstack.pop()
        left = numstack.pop()
        result = left - right
        numstack.append(result)
    elif op == "*":
        right = numstack.pop()
        left = numstack.pop()
        result = left * right
        numstack.append(result)
    elif op == "/":
        right = numstack.pop()
        left = numstack.pop()
        result = left / right
        numstack.append(result)
    else:
        print("op is invalid")



######################################

# Program Running

# stack implemented with lists
# numstack will contain numbers of the infix expression
# opstack will contain the operations
numstack = []
opstack = []

# Given Case
# given in-fix expression
givenInfix = "10 * 2 - 15"
print("Given Infix:", givenInfix)
calculatorExpression(givenInfix)

# Invalid Case
invalidInfix = "9- +9"
print("Invalid Infix:", invalidInfix)
calculatorExpression(invalidInfix)

# Complex Case
complexInfix = "98 + 7 * 2"
print("Complex Infix:", complexInfix)
calculatorExpression(complexInfix)
######################################

# Time and Space Complexity
"""
Time Complexity
I'm pretty sure it is O(n) time.

string.find(substring) is O(n x m) where n is length of string, and m is length of substring.
i have a str.find(" "), which is n is length of str (i.e 10) and m is length of " " which is 1
    but it basically becomes n + n + n + ... which might be O(3n) which drops the coefficent down to O(n) time


Space Complexity
I believe my solution is O(n) space.

the infix expression i give to my program is a set string that never changes. N
i create two stacks, which usually hold less the the given input. maybe like +3
i sometimes create temporary variables as i move elements out of my stacks. those 3 elements are just stored elsewhere, so it doesn't change space.
then that 3 variables are put back into the stack as 1 result. -3. +1. so net -2
at the end, the last number is printed out to the console

by the end, it is N +3 -2 == N + 1
and that is basically just N
"""

# Comment how it would be different if implemented with deque
"""
A deque is a generalization of stacks and queues.
You can add values to either end of the deque, and pop from either end of the deque.
Implementing this in a deque, we can take advantage of the duel ends for our stacks.

Right now in lists, i've made a number stack, and an operator stack, using two lists.
in a deque, i can insert the numbers the front of the deque, and append the operators in the back of the deque
I can quickly access the numbers and operators from the respective ends of the deque.
"""

""""""
```

```python
# Lab 8
# CSC 2720
# Anshuk Arun

# Simple Calculator using deque

#Collections.deque data structure in python
# design calculator
# not use in-built math functions

# post-fix expression
# 3 5 + 1 -

from collections import deque

# calculate function
def calc(expression=""):

    # create the deque
    dq = deque()

    # use the front part of dq for operands / numbers
    # use the back part of dq for operations / symbols

    # expression list
    exprList = expression.split(' ')
    # index of the string list
    c = 0

    # only run the given expression
    while c < len(exprList):
        
        # get the next element in the expression list
        val = exprList[c]
        # increment the index
        c = c + 1

        # Number. add to front of deque
        if val.isnumeric():
            dq.appendleft(int(val))
        # Operator. add to back of deque. Attempt to do operation.
        elif val == "+" or val == "-" or val == "*" or val == "/":
            # add the operation to the stack
            dq.append(val)
            
            # check if there is enough numbers to do a calculation
            # at least two numbers in the front end of the deque for a calculation
            # numbers are type int, operators are type string
            if type(dq[0]) == type(dq[1]):
                doOperation(dq, val)
        # Invalid Element
        else:
            return "Invalid Expression"
            
    # return the final result
    return dq[0]

# with two numbers and an operator, get the result of the operation
def doOperation(deq, valstr):
    # do the operation
    if valstr == "*":
        # multiply the previous two numbers
        result = deq.popleft() * deq.popleft()
        # push result (multiplication) into number side of deque
        deq.appendleft(result)
        # remove the used operation *
        deq.pop()
    elif valstr == "/":
        # divide in the correct order
        v2 = deq.popleft()
        v1 = deq.popleft()
        result = v1 / v2
        # push result (division) into number deque
        deq.appendleft(result)
        # remove used operation /
        deq.pop()
    elif valstr == "+":
        # add previous two numbers
        result = deq.popleft() + deq.popleft()
        # push result (addition) into number deque
        deq.appendleft(result)
        # remove used operation +
        deq.pop()
    elif valstr == "-":
        v2 = deq.popleft()
        v1 = deq.popleft()
        result = v1 - v2
        # push result (subtraction) into number side of deque
        deq.appendleft(result)
        # remove the used operation -
        deq.pop()


##########################################

# calculate simple expression
# regular case
# 3 5 + 1 -
simple = "3 5 + 1 -"
print(f"Simple \"{simple}\": ", calc(simple))

# edge case
edge = "9- Null + 9"
print(f"Edge \"{edge}\": ", calc(edge))

# complex case
complex = "9 8 7 + 4 * + 3 -"
print(f"Complex \"{complex}\": ", calc(complex))
# should = 66

###################################333

# Time Complexity
"""
This solution is O(n) time.
It has a while loop, iterating once over the input, and does not have to repeat iterations.
"""
# Space Complexity
"""
This solution is O(n) space.
the initial input has n elements.
the deque fills with some elements, usually 3, before a doOperation() is called and the deque reduces down to 1 element.
such, N for inital input + deque (N+1). This would be 2N +1, which is just N in big O.
"""

"""End of File"""
```

```python
# Final
# Anshuk Arun
# Problem 2

from collections import deque

# post-fix expression as input
givenPF = ["2", "1", "+", "4", "*"]

# Evaluates a Post-Fix Expression
def evalPF(pfexp):

    dq = deque()
    # Deque to act as a double-ended stack
    # front for operands/numbers
    # back for operators/symbols

    # run the expression once
    for i in range(len(pfexp)):

        # Current Element in expression
        val = pfexp[i]

        # Number? Front of DQ
        if val.isnumeric():
            dq.appendleft(int(val))
        # Operator? Back of DQ
        elif val == "+" or val == "-" or val == "*" or val == "/":
            dq.append(val)
            
            # at least two numbers, do the operation
            if type(dq[0]) == type(dq[1]):
                # do operation, but also check for division by 0
                if operation(dq) == "ERROR: Division by 0":
                    return "ERROR: Division by 0"
        # Invalid Element
        else:
            return "Invalid Expression"

    # return final result
    return dq[0]

# helper function for evaluating post-fix expresion
# does an operation
def operation(dq):
    # get the operator from back of dq
    op = dq.pop()
    # result
    res = 0
    
    # based on operator, do the corresponding operation
    if op == "*":
        # multiply the previous two numbers
        res = dq.popleft() * dq.popleft()
        
    elif op == "/":
        # divide, accounting for order
        v2 = dq.popleft()
        v1 = dq.popleft()

        # Divide by 0?
        if v2 == 0:
            return "ERROR: Division by 0"
        # Division
        else:
            res = v1 / v2
        
    elif op == "+":
        # add the previous two numbers
        res = dq.popleft() + dq.popleft()
        
    elif op == "-":
        # subtract the previous two numbers, accounting for order
        v2 = dq.popleft()
        v1 = dq.popleft()
        res = v1 - v2

    # push result into number (front) side of deque
    dq.appendleft(res)



# evaluate post-fix expression in O(n) time and O(n) space complexity
# space complexity from a data structure which helps solve this problem
# solution with not correct space complexity, half credit

# ================
# given case
print("Final Result:", evalPF(givenPF))

# Time Complexity: O(N)
# iteration over post-fix expression once

# Space Complexity: O(N)
# space complexity from the deque structure that helps evaluate expression
```
